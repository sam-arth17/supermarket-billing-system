# Supermarket Billing System – Transaction Retrieval

This notebook retrieves and displays previously saved supermarket transaction records from the `transactions.pkl` file using Python's Pickle module.

### Features
- Load saved transaction records
- Display total number of transactions
- View individual bill summaries
- Retrieve item-wise purchase details

```python
## 1. Loading Saved Transactions
import pickle

with open("transactions.pkl", "rb") as file:
    transactions = pickle.load(file)
```

```python
## 2. Displaying Total Transactions
print("Total Transactions:", len(transactions))
```

```python
## 3. Displaying Transaction Summaries

for transaction in transactions:
    print("\nBill Number:", transaction["Bill Number"])
    print("Customer Name:", transaction["Customer Name"])
    print("Subtotal:", transaction["Subtotal"])
    print("Item-wise Discount:", transaction["Item-wise Discount"])
    print("Additional Discount:", transaction["Additional Discount Amount"])
    print("Final Payable:", transaction["Final Payable"])
    print("-" * 40)
```

```python
## 4. Retrieving Individual Item Details
for transaction in transactions:
    print("\nBill Number:", transaction["Bill Number"])
    print("Customer Name:", transaction["Customer Name"])

    for item in transaction["Items"]:
        print(
            item["Name"],
            item["Category"],
            item["Quantity"],
            item["Price"],
            item["Discount Amount"]
        )
```

```python
import pickle

transactions = []

with open("transactions.pkl", "wb") as file:
    pickle.dump(transactions, file)

print("All transaction records deleted successfully!")
```
