# Supermarket Billing System

A Python-based supermarket billing application that generates customer bills, applies item-wise and additional bill discounts and stores transaction records using Pickle file handling.

### Features
- Multiple items per transaction
- Category-based discounts
- Additional discount on bills above ₹1,000
- Input validation
- Formatted customer receipts
- Persistent transaction storage using Pickle

### Discount Rules

| Category / Condition | Discount |
|---|---:|
| Dairy | 5% |
| Fruits | 10% |
| Other | 0% |
| Bill subtotal above ₹1,000 | Additional 5% |

```python
import pickle
try:
    with open("transactions.pkl", "rb") as file:
        transactions = pickle.load(file)

except FileNotFoundError:
    transactions = []

```

## 2. Customer Details and Initialisation

```python
bill_number = len(transactions) + 1


name=input("Enter customer name: ")
items=[]
total=0
subtotal=0
total_discount=0
```

## 3. Item Entry and Input Validation

```python
while True:
    
    item=input("Enter item name: ")
    while True:
        try:
            price = int(input("Enter price: "))
            quantity = int(input("Enter quantity: "))

            if price > 0 and quantity > 0:
                break
            else:
                print("Price and quantity must be greater than zero.")

        except ValueError:
            print("Please enter valid numbers.")
    while True:
        category = input("Enter category (dairy/fruits/other): ").lower()

        if category == "dairy" or category == "fruits" or category == "other":
            break
        else:
            print("Invalid category. Please try again.")
    amount = quantity * price

    if category == "dairy":
        discount = 5
    elif category == "fruits":
        discount = 10
    else:
        discount = 0

    
    discount_amount = (discount*amount)/100
    final_amount =amount - discount_amount
    item_details={
        "Name": item,
        "Price": price,
        "Category": category,
        "Quantity": quantity,
        "Discount": discount,
        "Discount Amount": discount_amount,
        "Original Amount": amount,
        "Final Amount": final_amount
        
    }
    items.append(item_details)

    subtotal+=amount
    total += final_amount
    total_discount += discount_amount
    while True:
        check = input("Is there any other item? Yes/No: ").lower()

        if check == "yes" or check == "no":
            break
        else:
            print("Please enter Yes or No.")
    if check == "yes":
        continue
    else:
        break
```

## 4. Discount Calculation

```python
if subtotal > 1000:
    additional_discount = 5
else:
    additional_discount = 0

bill_discount = (additional_discount * subtotal) / 100

final_payable = total - bill_discount
```

## 5. Receipt Generation

```python
print("\n", "=" * 89)
print("\n", " " * 35, "CUSTOMER BILL")
print("\n", "=" * 89)

print("Bill Number:", bill_number)
print("\nCustomer Name:", name)
print("-" * 89)


print(f"{'Item':<15}{'Category':<15}{'Qty':<7}{'Price':<10}{'Original':<12}{'Discount':<12}{'Final':<10}")
print("-" * 89)


for i in items:

    print(
        f"{i['Name']:<15}"
        f"{i['Category']:<15}"
        f"{i['Quantity']:<7}"
        f"{i['Price']:<10.2f}"
        f"{i['Original Amount']:<12.2f}"
        f"{str(i['Discount']) + '%':<12}"
        f"{i['Final Amount']:<10.2f}"
    )

print("-" * 89)

print()
print("Subtotal =", round(subtotal, 2))
print("Item-wise Discount =", round(total_discount, 2))
print(f"Additional Bill Discount ({additional_discount}%) =", round(bill_discount, 2))
print("Final Payable Amount =", round(final_payable, 2))

print("\n", "=" * 30, "THANK YOU FOR SHOPPING", "=" * 30)
```

## 6. Saving Transaction Records

```python
transaction = {
    "Bill Number": bill_number,
    "Customer Name": name,
    "Items": items,
    "Subtotal": subtotal,
    "Item-wise Discount": total_discount,
    "Additional Discount Percentage": additional_discount,
    "Additional Discount Amount": bill_discount,
    "Final Payable": final_payable
}

transactions.append(transaction)
with open("transactions.pkl", "wb") as file:
    pickle.dump(transactions, file)

print("\nTransaction saved successfully!")


```
